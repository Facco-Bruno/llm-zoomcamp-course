```python
from fastembed.embedding import TextEmbedding
import numpy as np
from numpy.linalg import norm

embed_model = TextEmbedding(model_name="jinaai/jina-embeddings-v2-small-en")

query = "I just discovered the course. Can I join now?"
doc = "Can I still join the course after the start date?"

query_embedding = next(embed_model.embed([query]))
doc_embedding = next(embed_model.embed([doc]))

# Similaridade do cosseno
cos_sim = np.dot(query_embedding, doc_embedding) / (norm(query_embedding) * norm(doc_embedding))
print(f"Cosine similarity: {cos_sim:.2f}")
```

```python
from fastembed import TextEmbedding
from qdrant_client import QdrantClient
from qdrant_client.models import PointStruct, VectorParams, Distance
import numpy as np
import requests
import uuid

# Baixa os documentos
docs_url = 'https://github.com/alexeygrigorev/llm-rag-workshop/raw/main/notebooks/documents.json'
documents_raw = requests.get(docs_url).json()

# Filtra os documentos do curso machine-learning-zoomcamp
documents = []
for course in documents_raw:
    if course['course'] == 'machine-learning-zoomcamp':
        for doc in course['documents']:
            text = doc['question'] + ' ' + doc['text']
            documents.append({
                "id": str(uuid.uuid4()),
                "text": text
            })

# Inicializa o modelo (dimensão 384)
model = TextEmbedding(model_name="BAAI/bge-small-en-v1.5")

# Inicializa o Qdrant (em memória)
qdrant = QdrantClient(":memory:")
dim = 384

# Cria a coleção
qdrant.recreate_collection(
    collection_name="ml_zoomcamp_faq",
    vectors_config=VectorParams(size=dim, distance=Distance.COSINE)
)

# Embedding + upload dos documentos
points = []
for doc in documents:
    vector = next(model.embed([doc["text"]]))
    points.append(PointStruct(id=doc["id"], vector=vector, payload={"text": doc["text"]}))

qdrant.upsert(collection_name="ml_zoomcamp_faq", points=points)

# Query (mesma da Q1)
query = "I just discovered the course. Can I join now?"
query_vector = next(model.embed([query]))

# Pesquisa
results = qdrant.search(
    collection_name="ml_zoomcamp_faq",
    query_vector=query_vector,
    limit=1
)

# Mostra o score do melhor resultado
print("✅ Highest score:", results[0].score)
```
